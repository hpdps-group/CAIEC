```python
from compressai import gpu_codec
import nvidia.nvcomp as nvcomp
import numpy as np
import torch
import time

path = "/hwj/project/aiz-accelerate/data/nyx-dark_matter_density.npy"

# 1) load npy -> tensor
arr = np.load(path)
x = torch.from_numpy(arr).float().to(device)
print(x.shape, x.dtype, x.device)

# 2) build codec
cfg = {
    "normalizer": {"name": "minmax", "params": {"scale_bits": 8}},
    "quantizer": {"name": "eb", "params": {"eb": 0.8}},
    "algo": {"name": "bitcomp", "params": {}},
}
codec = gpu_codec.build_gpu_codec(cfg)

for _ in range(5):
    ga_output = net.g_a(x)
    pack = codec.compress(ga_output)

num_iters = 10
t0 = time.perf_counter()
for _ in range(num_iters):
    ga_output = net.g_a(x)
torch.cuda.synchronize()
t1 = time.perf_counter()

t2 = time.perf_counter()
for _ in range(num_iters):
    pack = codec.compress(ga_output)
torch.cuda.synchronize()
t3 = time.perf_counter()

strings = pack["strings"]
state   = pack["state"]

compressed_bytes = sum(len(s) for s in strings)
print("Compressed size (bytes):", compressed_bytes)

# 原有 Lossless CR：按 latent float32 大小算
raw_latent_bytes = ga_output.numel() * ga_output.element_size()
print("Lossless Compression Ratio:",
      raw_latent_bytes / compressed_bytes)

print("Total Compression Ratio:", x.numel() * x.element_size() / compressed_bytes)

# 原有 bpp：按 input 图像/体素大小算
bpp = compressed_bytes * 8 / (x.shape[2] * x.shape[3] * x.shape[0])
print("Total Bits per pixel (bpp):", bpp)

total_time = t1 - t0 + t3 - t2
inference_time = t1 - t0
code_time = t3 - t2
avg_time = total_time / num_iters

# 吞吐量口径：按 input x 的字节数
input_bytes = x.numel() * x.element_size()

gb_per_sec = (input_bytes * num_iters) / total_time / (1024 ** 3)
infe_per_sec = (input_bytes * num_iters) / inference_time / (1024 ** 3)
codec_per_sec = (raw_latent_bytes * num_iters) / code_time / (1024 ** 3)

print(f"[E2E Compress] throughput: {gb_per_sec:.3f} GB/s ")
print(f"[E2E Compress Inference] throughput: {infe_per_sec:.3f} GB/s ")
print(f"[E2E Compress Codec] throughput: {codec_per_sec:.3f} GB/s ")
```

```python
# -----------------------------------------
# 3) Decompression 测试
# -----------------------------------------
num_iters = 10
print(state)
# b = strings[0]
# arr = nvcomp.as_array(b).cuda()
# decomp_arr = codec.algo.codec.decode(arr)
# y_hat_bytes = torch.utils.dlpack.from_dlpack(decomp_arr)
# y_q = y_hat_bytes.view(torch.int16)
y_hat = codec.decompress(strings, state)

# pack 中已有：
#   strings = pack["strings"]
#   state   = pack["state"]
# torch.cuda.synchronize()
# warm-up
for _ in range(5):
    y_hat = codec.decompress(strings, state)
    torch.cuda.synchronize()

# torch.cuda.synchronize()

# 正式计时
t0 = time.perf_counter()
for _ in range(num_iters):
    y_hat = codec.decompress(strings, state)
torch.cuda.synchronize()
t1 = time.perf_counter()

total_time = t1 - t0
avg_time = total_time / num_iters

# -----------------------------------------
# 输出吞吐量 (按 latent 大小，或按 input 大小)
# -----------------------------------------
latent_bytes = ga_output.numel() * ga_output.element_size()

samples_per_sec = num_iters / total_time
gb_per_sec = (latent_bytes * num_iters) / total_time / (1024 ** 3)

print(f"\n[E2E Decompress] throughput: {samples_per_sec:.3f} samples/s")
print(f"[E2E Decompress] throughput: {gb_per_sec:.3f} GB/s")

# -----------------------------------------
# 可选：检查解压结果尺寸
# -----------------------------------------
print("Decompressed latent shape:", y_hat.shape)

# -----------------------------------------
# 可选：如果你希望通过解码器得到最终重建图像
# -----------------------------------------
y_hat = y_hat.reshape(ga_output.shape)
recon = net.g_s(y_hat)   # 若你有 g_s
print("Reconstruction shape:", recon.shape)

# -----------------------------------------
# 可选：误差检查（PSNR / MSE）
# -----------------------------------------
mse = torch.mean((ga_output - y_hat) ** 2).item()
print("Latent MSE:", mse)

if 'recon' in locals():
    mse_recon = torch.mean((x - recon) ** 2).item()
    print("Final Reconstruction MSE:", mse_recon)
    
# ---------- 显式释放 ----------
del pack, strings, state
del ga_output
del x, arr
del codec
del net
# 如果 net 是临时用的，也可以 del net

torch.cuda.synchronize()
torch.cuda.empty_cache()
```

### Compute metrics

Now, let's compute some common metrics...

```python
def compute_psnr(a, b):
    mse = torch.mean((a - b)**2).item()
    return -10 * math.log10(mse)

def compute_msssim(a, b):
    return ms_ssim(a, b, data_range=1.).item()

def compute_bpp(out_net):
    size = out_net['x_hat'].size()
    num_pixels = size[0] * size[2] * size[3]
    return sum(torch.log(likelihoods).sum() / (-math.log(2) * num_pixels)
              for likelihoods in out_net['likelihoods'].values()).item()
```

```python
print(f'PSNR: {compute_psnr(x, out_net["x_hat"]):.2f}dB')
print(f'MS-SSIM: {compute_msssim(x, out_net["x_hat"]):.4f}')
print(f'Bit-rate: {compute_bpp(out_net):.3f} bpp')
```

## 2. Comparison to classical codecs

Let's perform some comparison against JPEG and WebP as they are included in the Pillow library.

```python
def pillow_encode(img, fmt='jpeg', quality=10):
    tmp = io.BytesIO()
    img.save(tmp, format=fmt, quality=quality)
    tmp.seek(0)
    filesize = tmp.getbuffer().nbytes
    bpp = filesize * float(8) / (img.size[0] * img.size[1])
    rec = Image.open(tmp)
    return rec, bpp
```

```python
def find_closest_bpp(target, img, fmt='jpeg'):
    lower = 0
    upper = 100
    prev_mid = upper
    for i in range(10):
        mid = (upper - lower) / 2 + lower
        if int(mid) == int(prev_mid):
            break
        rec, bpp = pillow_encode(img, fmt=fmt, quality=int(mid))
        if bpp > target:
            upper = mid - 1
        else:
            lower = mid
    return rec, bpp

def find_closest_psnr(target, img, fmt='jpeg'):
    lower = 0
    upper = 100
    prev_mid = upper
    
    def _psnr(a, b):
        a = np.asarray(a).astype(np.float32)
        b = np.asarray(b).astype(np.float32)
        mse = np.mean(np.square(a - b))
        return 20*math.log10(255.) -10. * math.log10(mse)
    
    for i in range(10):
        mid = (upper - lower) / 2 + lower
        if int(mid) == int(prev_mid):
            break
        prev_mid = mid
        rec, bpp = pillow_encode(img, fmt=fmt, quality=int(mid))
        psnr_val = _psnr(rec, img)
        if psnr_val > target:
            upper = mid - 1
        else:
            lower = mid
    return rec, bpp, psnr_val

def find_closest_msssim(target, img, fmt='jpeg'):
    lower = 0
    upper = 100
    prev_mid = upper
    
    def _mssim(a, b):
        a = torch.from_numpy(np.asarray(a).astype(np.float32)).permute(2, 0, 1).unsqueeze(0)
        b = torch.from_numpy(np.asarray(b).astype(np.float32)).permute(2, 0, 1).unsqueeze(0)
        return ms_ssim(a, b, data_range=255.).item()

    for i in range(10):
        mid = (upper - lower) / 2 + lower
        if int(mid) == int(prev_mid):
            break
        prev_mid = mid
        rec, bpp = pillow_encode(img, fmt=fmt, quality=int(mid))
        msssim_val = _mssim(rec, img)
        if msssim_val > target:
            upper = mid - 1
        else:
            lower = mid
    return rec, bpp, msssim_val
```
